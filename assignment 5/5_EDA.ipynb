```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/Cardiotocographic.csv')
df
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.fillna(df.mean(),inplace=True)
```

```python
df.dtypes
```

```python
#to find outliers (IQR method)

Q1 = np.percentile(df, 25, axis=0)
Q3 = np.percentile(df, 75, axis=0)

IQR = Q3 - Q1

outliers = ((df < (Q1 - 1.5*IQR)) | (df > (Q3 + 1.5*IQR)))
outliers.sum()
```

Data Visualization

```python
# HISTOGRAM

df.hist(figsize=(12,10))
plt.show()
```

```python
#BOX PLOT

plt.figure(figsize=(12,5))
sns.boxplot(data=df)
plt.yscale("log")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.show()
```

```python
#SCATTER PLOT
sns.scatterplot(x="LB", y="MSTV", data=df)
plt.show()
```

```python
#PAIR PLOT

sns.pairplot(df, plot_kws={'alpha':0.3})
plt.show()
```

```python
#VIOLIN PLOT
sns.violinplot(data=df)
plt.xticks(rotation=90)
plt.show()
```
