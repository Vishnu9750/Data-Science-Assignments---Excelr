5.	Remove duplicate patient records based on PatientID.

```python
Patient_data.drop_duplicates(subset=['PatientID'], inplace=True)
print("Duplicate patient records removed based on PatientID.")
print(Patient_data)
```

6.	Fill missing BillAmount values with the mean bill amount.

```python
mean_bill = patients_billing['BillAmount'].mean()
patients_billing['BillAmount'].fillna(mean_bill)
```

7.	Merge the billing dataset with patient dataset on PatientID.

```python
merged_data = pd.merge(Patient_data, Billing_data, on='PatientID')
print("Merged DataFrames:")
print(merged_data.head())
```

8.	Concatenate an additional DataFrame that contains new patients for the current week (row-wise).

```python
new_patients_data = pd.DataFrame({
    'PatientID': [106, 107],
    'Name': ['Frank', 'Grace'],
    'Department': ['Orthopedics', 'Cardiology'],
    'Doctor': ['Dr. Lee', 'Dr. Smith'],
    'BillAmount': [8000.0, 5500.0]
})

Patient_data = pd.concat([Patient_data, new_patients_data], ignore_index=True)
print(Patient_data)
```

9.	Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).

```python
patient_data_with_billing = pd.merge(
    Patient_data,
    Billing_data[['PatientID', 'InsuranceCovered', 'FinalAmount']],on='PatientID'
)

print(patient_data_with_billing.head(10))
```
