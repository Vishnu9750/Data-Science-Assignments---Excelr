```python
import math
from scipy.stats import norm
```

```python
# Given values
x_bar   = 3050     # sample mean
mean    = 600       # mean units
sigma_x = 25     # std dev units
n       = 25           # sample size
alpha   = 0.05     # significance level

print("Sample Mean  :", x_bar)
print("Mean units   :", mean)
print("Std dev units:", sigma_x)
print("Sample size  :", n)
```

```python
mean_theoretical  = 1000+5*mean
print("Theroretical mean  = " , mean_theoretical)
```

```python
sigma = 5*sigma_x
print("standard deviation = " , sigma)
```

```python
#hypothesis statement
print("H0: μ = 4000")
print("H1: μ > 4000 (Right-tailed test)")
```

```python
# Z formula

z = (x_bar - mean_theoretical) / (sigma / math.sqrt(n))
print("Z Test Statistic:", z)
```

```python
# Critical value

z_critical = norm.ppf(1 - alpha)
print("Critical Z Value:", z_critical)
```

```python
if z > z_critical:
    print("Decision: Reject Null Hypothesis")
else:
    print("Decision: Fail to Reject Null Hypothesis")
```
