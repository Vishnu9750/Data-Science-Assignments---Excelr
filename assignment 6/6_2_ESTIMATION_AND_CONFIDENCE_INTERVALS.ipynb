```python
import numpy as np
import math
from scipy.stats import t,norm
```

```python
# Sample data
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07,
        1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

n = len(data)
confidence = 0.99
alpha = 1 - confidence

print("Sample size:", n)
```

```python
# Sample statistics
mean = np.mean(data)
std = np.std(data, ddof=1)   # sample std dev

print("Sample Mean:", mean)
print("Sample Standard Deviation:", std)
```

99% CI using Sample Std Dev (t-distribution)

```python
# Degrees of freedom
df = n - 1

t_critical = t.ppf(1 - alpha/2, df)

print("Degrees of freedom:", df)
print("t critical value:", t_critical)
```

```python
# Margin of error

ME_t = t_critical * (std / math.sqrt(n))
print("Margin of Error (t):", ME_t)
```

```python
lower_t = mean - ME_t
upper_t = mean + ME_t

print("99% CI using t-distribution for :- ")
print("Lower bound:", lower_t)
print("Upper bound:", upper_t)
```

99% CI using Known Population Std Dev

```python
sigma = 0.2   # population std dev
```

```python
z_critical = norm.ppf(1 - alpha/2)
print("Z critical value:", z_critical)
```

```python
ME_z = z_critical * (sigma / math.sqrt(n)) #margin error
print("Margin of Error (Z):", ME_z)
```

```python
lower_z = mean - ME_z
upper_z = mean + ME_z

print("99% CI using Z-distribution for :-")
print("Lower bound:", lower_z)
print("Upper bound:", upper_z)
```
