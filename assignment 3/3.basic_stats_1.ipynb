Descriptive Analytics for Numerical Columns

```python
import pandas as pd
```

```python
df = pd.read_csv("/content/sales_data_with_discounts.csv")
```

```python
df
```

```python
df.info()  # to see information of the data
```

```python
df.columns   #TO see all the Column name s
```

```python
X_nume = df[['Volume', 'Avg Price', 'Total Sales Value', 'Discount Rate (%)', 'Discount Amount', 'Net Sales Value']]
```

```python
X_nume
```

```python
# Mean
mean_value = X_nume.mean()
print("Mean Values:\n", mean_value)

# Median
median_value = X_nume.median()
print("\nMedian Values:\n", median_value)

# Mode
mode_value = X_nume.mode().iloc[0]
print("\nMode Values:\n", mode_value)

# Standard Deviation
std_value = X_nume.std()
print("\nStandard Deviation:\n", std_value)
```

#Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

#Histograms for numerical data

```python
for column in X_nume.columns:
    plt.figure(figsize=(8, 6))
    sns.histplot(X_nume[column], kde=True)
    plt.title(f'Histogram of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')
    plt.show()
```

#Box Plot for numerical data

```python
for column in X_nume.columns:
    plt.figure(figsize=(12, 6))
    sns.boxplot(y=X_nume[column] , width =0.6)
    plt.title(f'Boxplot of {column}')
    plt.ylabel(column)
    plt.show()
```

#Bar chart for categorical data

```python
X_cate = ['Day', 'SKU', 'City', 'BU', 'Brand', 'Model']

for col in X_cate:
    plt.figure(figsize=(10, 5))
    df[col].value_counts().plot(kind='bar')
    plt.title(f'Bar Chart of {col}')
    plt.xlabel(col)
    plt.ylabel('Count / Frequency')
    plt.show()
```
